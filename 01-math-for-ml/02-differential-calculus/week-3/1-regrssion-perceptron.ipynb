```python
import numpy as np
from sklearn.preprocessing import StandardScaler
```

```python
X_org = np.array([
    [1000, 2],
    [1500, 3],
    [2000, 4],
    [2500, 5]
])

y = np.array([
    [200],
    [300],
    [400],
    [500]
])
```

```python
# shape 
print(X_org.shape)
print(y.shape)
```

```python
scaler = StandardScaler()
```

```python
X = scaler.fit_transform(X_org)
```

```python
# initialize the weight and bias
def initialization(X):
    W = np.zeros(X.shape[1])
    W = W.reshape(-1,1)
    b = 0
    return (W,b)
```

```python
initialization(X)
```

### Forward Propagation
Calculates the predicted output:
$$ \hat{y} = XW + b $$

```python
def forward_propagtion(X,W,b):
    y_hat = X @ W + b
    return y_hat
```

### Mean Squared Error (MSE) Cost Function
Calculates the error between true labels and predictions:
$$ J(W,b) = \frac{1}{2m} \sum_{i=1}^{m} (y^{(i)} - \hat{y}^{(i)})^2 $$

```python
def cost(y_hat,y):
    m = len(y)

    cost = (1/(2*m)) * np.sum((y - y_hat)**2)
    
    return cost
```

### Gradients (Backward Propagation)
Calculates the partial derivatives of the cost function with respect to weights ($W$) and bias ($b$):
$$ \frac{\partial J}{\partial W} = -\frac{1}{m} X^T (y - \hat{y}) $$
$$ \frac{\partial J}{\partial b} = -\frac{1}{m} \sum_{i=1}^{m} (y^{(i)} - \hat{y}^{(i)}) $$

```python
def backward_propagation(X,y,y_hat):
    m = len(y)

    error = y - y_hat

    dw = -1 * (1/m) * (X.T @ error)

    db = -1* (1/m) * np.sum(error)

    return dw,db
```

### Gradient Descent Parameter Update
Updates the weights and bias using the learning rate ($\alpha$):
$$ W = W - \alpha \frac{\partial J}{\partial W} $$
$$ b = b - \alpha \frac{\partial J}{\partial b} $$

```python
# now let's use all this in creating single layer perceptron 
iteration = 1000
learning_rate = 0.01

# initialization of w & b
W,b = initialization(X)

for step in range(iteration):
    
    # Forward propagtion
    y_hat = forward_propagtion(X,W,b)

    # calcaulate Cost
    curr_cost = cost(y_hat,y)

    # Backward Propagation
    dw,db = backward_propagation(X,y,y_hat)

    # update parameter
    W = W - learning_rate*dw
    b = b - learning_rate*db

    if step%100 == 0:
        print(f"Current Iteratio is {step} and cost = {curr_cost}")
```
