```python
# let's create a simple dataset
from sklearn.datasets import make_blobs
import matplotlib.pyplot as plt
import numpy as np

X, y = make_blobs(
    n_samples=200,
    centers=2,
    n_features=2,
    random_state=42
)
```

```python
print(X.shape)
print(y.shape)
```

```python
plt.scatter(X[:, 0], X[:, 1], c=y)
plt.xlabel("X1")
plt.ylabel("X2")
plt.show()
```
